```python
from mpmath import mp
mp.dps = 60
```

```python
N = 101 
x = lambda i: -1 + mp.mpf(2*i) / mp.mpf(N - 1)

P = mp.matrix(N, N)
for l in range(N):
    for i in range(N):
        P[i, l] = mp.sqrt(2*l+1) * mp.legendre(l, x(i))

invP = P**-1

print(mp.mnorm((P @ invP) - mp.eye(N), p=mp.inf))
```
