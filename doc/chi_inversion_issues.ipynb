# Susceptibility inversion problem on the Keldysh contour

Solve
$$
\int_\mathcal{C} d\bar z \chi(z, \bar z) X(\bar z, z') = \delta_\mathcal{C}(z, z')
$$
with respect to $X(\bar z, z')$.

Keldysh components of $\chi$:

$$
\begin{align}
    \chi^{-+}(t, t') &= \chi^>(t, t') = \langle S_z(t) S_z(t')\rangle,\\
    \chi^{+-}(t, t') &= \chi^<(t, t') = \langle S_z(t') S_z(t)\rangle,\\
    \chi^{++}(t, t') &= \langle \mathtt{T} \langle S_z(t) S_z(t')\rangle
                      = \theta(t-t')\chi^{-+}(t,t') + \theta(t'-t)\chi^{+-}(t,t'),\\
    \chi^{--}(t, t') &= \langle \bar{\mathtt{T}} \langle S_z(t) S_z(t')\rangle
                      = \theta(t'-t)\chi^{-+}(t,t') + \theta(t-t')\chi^{+-}(t,t').\\
\end{align}
$$

# Discretization 1: $2N_t$ time slices on the unfolded contour

Let's fix $z = (a, t)$, $z' = (b, t')$ and introduce a real time grid on the unfolded Keldysh contour,
$$
    t_i = \left\{
    \begin{array}{ll}
        i\Delta t, &i \in [0, N_t-1],\\
        (2N_t-i-1)\Delta t, &i \in [N_t, 2N_t-1].
    \end{array}
    \right.
$$

$$
\begin{align}
\int_\mathcal{C} d\bar z \chi(z, \bar z) X(\bar z, z') &\approx
    \frac{\Delta t}{2} \left(\chi^{a+}(t, t_0) X^{+b}(t_0, t') +
                             \chi^{a+}(t, t_1) X^{+b}(t_1, t')\right)\\
    &+
    \frac{\Delta t}{2} \left(\chi^{a+}(t, t_1) X^{+b}(t_1, t') +
                             \chi^{a+}(t, t_2) X^{+b}(t_2, t')\right)\\
    &+\ldots\\
    &+
    \frac{\Delta t}{2} \left(\chi^{a+}(t, \Delta t(N_t-2)) X^{+b}(\Delta t(N_t-2), t') +
                             \chi^{a+}(t, \Delta t(N_t-1)) X^{+b}(\Delta t(N_t-1), t')\right)\\
\end{align}
$$
